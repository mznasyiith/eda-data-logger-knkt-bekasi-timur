# 03. Memuat dan Memeriksa Data (Importing and Inspecting the Data)

Notebook ini membaca salinan dataset yang telah diperiksa hak penggunaannya. Jangan menimpa berkas sumber.

```python
from pathlib import Path
import pandas as pd

from eda_knkt.config import PROCESSED_DATA_DIR
from eda_knkt.ingestion import load_csv, load_excel
from eda_knkt.validation import inspect_dataframe

# Dataset transkripsi Tabel 6–10 yang tersedia di repository.
DATA_FILE = "data_logger_knkt_tabel_6_10.csv"
data_path = PROCESSED_DATA_DIR / DATA_FILE

if not data_path.exists():
    raise FileNotFoundError(data_path)
suffix = data_path.suffix.lower()
if suffix == ".csv":
    df_raw = load_csv(data_path)
elif suffix in {".xlsx", ".xls"}:
    df_raw = load_excel(data_path)
else:
    raise ValueError(f"Format belum didukung: {suffix}")

print("Loaded:", data_path.name)
display(df_raw.head())
inspection = inspect_dataframe(df_raw)
print("Rows:", inspection["rows"])
print("Columns:", inspection["columns"])
display(pd.DataFrame({
    "column": inspection["column_names"],
    "dtype": [inspection["dtypes"][name] for name in inspection["column_names"]],
    "missing": [inspection["missing"][name] for name in inspection["column_names"]],
}))
```
