# BAB I. Pendahuluan dan Pemahaman Permasalahan (Understanding the Problem)

## 1.1 Latar belakang dan tujuan
Proyek ini melakukan *Exploratory Data Analysis* terhadap rekaman *Data Logger* yang dirujuk dalam Laporan Investigasi Kecelakaan Perkeretaapian KNKT Nomor **KNKT.26.04.03.02**, terkait tabrakan KA 4B Argo Bromo Anggrek dan KA 5568A Commuter Line di Stasiun Bekasi Timur pada 27 April 2026.

Tujuan analisis adalah memahami struktur, kualitas, karakteristik, distribusi, dan hubungan variabel yang benar-benar tersedia pada rekaman Data Logger. Proyek ini bukan investigasi independen, bukan penetapan penyebab kecelakaan, dan bukan penilaian tanggung jawab pihak mana pun.

## 1.2 Tipe masalah
Tahap ini merupakan EDA deskriptif/eksploratif, bukan tugas klasifikasi, regresi, atau klasterisasi yang telah ditetapkan. Penentuan model prediktif tidak menjadi prasyarat dan hanya dapat dibahas jika ada pertanyaan penelitian, target, serta data yang mendukung.

## 1.3 Metrik keberhasilan
Keberhasilan proyek EDA dinilai dari keterlacakan sumber, kelengkapan inspeksi, ketepatan perlakuan data, keterulangan proses, dan kejelasan temuan. Akurasi, F1-score, dan MAE tidak digunakan karena belum ada model prediktif yang menjadi ruang lingkup proyek.

## Batasan
- Analisis hanya mencakup variabel dan rentang waktu yang benar-benar terdapat dalam data.
- Tidak mengasumsikan arti, satuan, atau urutan waktu kolom tanpa verifikasi kamus data.
- Temuan deskriptif tidak membuktikan hubungan sebab-akibat.

## Catatan untuk laporan
Lengkapi identitas sumber, proses ekstraksi, versi dataset, cakupan waktu, dan pertanyaan eksplorasi setelah data aktual diperiksa. Jangan menulis jumlah baris/kolom atau hasil numerik sebelum inspeksi.
